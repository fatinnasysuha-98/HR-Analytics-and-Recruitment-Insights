# Question 2: AVERAGE SALARY DIFFERENCE (DIPLOMA VS MASTERS)

```python
avg_diploma = df[df['Education'] == 'Diploma']['Annual Income'].mean()
avg_masters = df[df['Education'] == "Masters"]['Annual Income'].mean()
difference = avg_masters - avg_diploma

print(f"Average Diploma Salary : RM {avg_diploma:,.2f}")
print(f"Average Master's Salary : RM {avg_masters:,.2f}")
print(f"Salary Difference : RM {difference:,.2f}")
```

# Question 3: AGE AFFECT SALARY FOR DIPLOMA HOLDERS

```python
# Filter Diploma Holder Only
diploma_df = df[df['Education'] == 'Diploma']
```

```python
# Correlation Analysis
correlation = diploma_df['Age'].corr(diploma_df['Annual Income'])

print(f"Correlation between Age and Income: {correlation:.4f}")
```

```python
# Regression Analysis
from scipy import stats

slope, intercept, r_value, p_value, std_err = stats.linregress(
    diploma_df['Age'],
    diploma_df['Annual Income']
)

print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"R^2: {r_value**2:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Regression eq: Income = {slope:.2f} x Age + {intercept:.2f}")
```

```python
# Filter High Income Earners
high_income = df[df['Annual Income'] >= 60000]
print(high_income[['Name', 'Annual Income', 'Phone']])
```

```python
# Most Common Phones
print(high_income['Phone'].value_counts())
```

```python
# Differentiate by Brands
phone_brands = high_income['Phone'].apply(
    lambda x: 'iPhone' if 'iPhone' in x else
    ('Nokia' if 'Nokia' in x else 'Other')
)

print(phone_brands.value_counts())
```
